### Question 3: Daily Savings Tracker

**Goal:**  
A student wants to calculate how much money they will have after saving the same amount every day for N days. Compute the total using a `while` loop.

**Requirements:**  
- Ask the user for the number of days (N) and the daily saving amount.  
- Use an accumulator variable to keep a running total.  
- Loop from 1 to N and add the daily amount each time.  
- Print the final total savings.

**Concepts:** Accumulator pattern, while loop with counter.

**Example:**
```
Enter number of days: 10
Enter daily saving amount: 50
Total savings after 10 days = 500
```


````python
days = int(input("Enter number of days: "))
daily_amount = float(input("Enter daily saving amount: "))

total_savings = 0           
day = 1
while day <= days:
    total_savings += daily_amount   
    day += 1

if total_savings == int(total_savings):
    total_savings = int(total_savings)
print("Total savings after", days, "days =", total_savings)
````

### Question 4: Even Seat Numbers in a Cinema

**Goal:**  
A cinema only sells tickets for even-numbered seats in a certain row. Print all even seat numbers between two limits given by the booking staff.

````python
start = int(input("Start seat: "))
end = int(input("End seat: "))

if start > end:
    start, end = end, start

seat = start
while seat <= end:
    if seat % 2 == 0:        
        print(seat)
    seat += 1
````

### Question 5: Online Banking Login Attempts

**Goal:**  
Create a simple online banking login simulation that allows a maximum of 3 password attempts before locking the account.

**Requirements:**  
- The correct password is the string `"secure@2024"` (hard-code it).  
- Use a `while` loop that continues as long as attempts remain and the password is incorrect.  
- After each wrong attempt, show how many attempts are left.  
- If the user enters the correct password, print `Login successful. Welcome to your account.`.  
- If all 3 attempts fail, print `Account temporarily locked for security reasons.`.

**Concepts:** Counter-controlled loop, conditional exit, user input validation.

**Example (successful):**
```
Enter password: hello
Incorrect. 2 attempts remaining.
Enter password: secure@2024
Login successful. Welcome to your account.
```

**Example (failed):**
```
Enter password: a
Incorrect. 2 attempts remaining.
Enter password: b
Incorrect. 1 attempt remaining.
Enter password: c
Account temporarily locked for security reasons.
```


````python
correct_password = "secure@2024"
attempts_left = 3
logged_in = False

while attempts_left > 0 and not logged_in:
    password = input("Enter password: ")
    if password == correct_password:
        logged_in = True
    else:
        attempts_left -= 1
        if attempts_left > 0:
            if attempts_left == 1:
                print("Incorrect. 1 attempt remaining.")
            else:
                print("Incorrect.", attempts_left, "attempts remaining.")

if logged_in:
    print("Login successful. Welcome to your account.")
else:
    print("Account temporarily locked for security reasons.")
````

### Question 7: Reversing a Product Serial Number

**Goal:**  
Quality control sometimes needs the reverse of a product serial number for internal checks. Reverse the digits of a positive integer using a `while` loop (no string conversion).

**Requirements:**  
- Read a positive integer (serial number).  
- Build the reversed number by repeatedly taking the last digit and appending it to a new number.  
- Print the reversed serial number.

**Concepts:** while loop, digit extraction, mathematical construction of a number.

**Example:**
```
Enter serial number: 12345
Reversed serial number: 54321
```


````python
serial = int(input("Enter serial number: "))

reversed_serial = 0
while serial > 0:
    last_digit = serial % 10                         
    reversed_serial = reversed_serial * 10 + last_digit  
    serial = serial // 10                            

print("Reversed serial number:", reversed_serial)
````

### Question 8: Customer Feedback Letter Counter

### Question 13: Product ID Primality Check (Security Code)

**Goal:**  
Some security systems treat a product ID as stronger if it is a prime number. Determine whether a given ID is prime using a `while` loop and the `else` clause of the loop.

**Requirements:**  
- Ask the user for an integer greater than 1 (the product ID).  
- Check divisibility from 2 up to the square root of the number (or up to n-1).  
- If any divisor is found, use `break` to exit the loop.  
- Use the `else` block of the `while` loop to print that the ID is prime (the else runs only when the loop was **not** broken).  
- If a divisor was found, print that the ID is not prime.

**Concepts:** while-else, break, primality testing logic.

**Example (prime):**
```
Enter product ID: 29
29 is a prime ID (strong security code).
```

**Example (not prime):**
```
Enter product ID: 15
15 is not a prime ID.
```


````python
product_id = int(input("Enter product ID: "))

divisor = 2
while divisor * divisor <= product_id:   
    if product_id % divisor == 0:
        print(product_id, "is not a prime ID.")
        break                            
    divisor += 1
else:
    print(product_id, "is a prime ID (strong security code).")
````

### Question 14: School Timetable Printer (Skip Assembly Day)

**Goal:**  
Print a simple multiplication-style timetable for periods 1 to 5, but skip the entire row when the period number is 3 (assembly time). Use nested loops and `continue`.

**Requirements:**  
- Outer loop controls the period number (1 to 5). Use a `while` loop.  
- Inner loop prints a sample line for each of 1 through 10 (you can print `Period X × Slot Y = result`).  
- When the outer counter equals 3, skip that whole row using `continue`.  
- Format the output clearly.

**Concepts:** Nested while/for, continue in outer loop context.

**Expected behavior:** Rows for periods 1, 2, 4 and 5 appear; the row for period 3 is completely omitted.


````python
period = 0
while period < 5:
    period += 1
    if period == 3:          
        continue
    print("--- Period", period, "---")
    for slot in range(1, 11):
        print("Period", period, "\u00d7 Slot", slot, "=", period * slot)
    print()
````

### Question 15: ATM Withdrawal Simulator

**Goal:**  
Simulate a basic ATM that allows a customer to withdraw money until the account balance is insufficient or the user chooses to exit.

**Requirements:**  
- Start with a hard-coded balance of 5000.  
- Display a simple menu repeatedly using a `while True` loop:  
  1. Check Balance  
  2. Withdraw  
  3. Exit  
- For withdrawal: ask the amount; if it is greater than the balance, print an error and do not change the balance; otherwise subtract it and show the new balance.  
- Exit the loop cleanly when the user selects option 3.  
- Use `break` for the exit option.

**Concepts:** Infinite loop with controlled exit, menu-driven program, conditional updates.

**Sample Interaction:**
```
1. Check Balance
2. Withdraw
3. Exit
Choose: 1
Current balance: 5000
Choose: 2
Amount to withdraw: 1200
Withdrawal successful. New balance: 3800
Choose: 3
Thank you for using the ATM.
```


````python
balance = 5000

while True:
    print("1. Check Balance")
    print("2. Withdraw")
    print("3. Exit")
    choice = input("Choose: ")

    if choice == "1":
        print("Current balance:", balance)
    elif choice == "2":
        amount = float(input("Amount to withdraw: "))
        if amount <= 0:
            print("Please enter a positive amount.")
        elif amount > balance:
            print("Error: Insufficient balance.")
        else:
            balance -= amount
            if balance == int(balance):
                balance = int(balance)
            print("Withdrawal successful. New balance:", balance)
    elif choice == "3":
        print("Thank you for using the ATM.")
        break
    else:
        print("Invalid choice. Please select 1, 2 or 3.")
````

### Question 16: Classroom Quiz Score Collector

**Goal:**  
A teacher collects quiz scores until she enters a sentinel value, then the program computes class statistics.

**Requirements:**  
- Repeatedly ask for a student score (0–100).  
- Stop collecting when the teacher enters `-1`.  
- Keep track of: total number of valid scores, sum of scores, highest score, and lowest score.  
- After the loop, if at least one score was entered, print:  
  - Number of students  
  - Average score (rounded to 2 decimal places)  
  - Highest and lowest scores  
- If no scores were entered, print an appropriate message.

**Concepts:** Sentinel-controlled loop, multiple accumulators, min/max tracking.

**Example:**
```
Enter score (-1 to finish): 85
Enter score (-1 to finish): 92
Enter score (-1 to finish): 78
Enter score (-1 to finish): -1
Students: 3
Average: 85.00
Highest: 92
Lowest: 78
```


````python
count = 0
total = 0
highest = 0
lowest = 100

while True:
    score = int(input("Enter score (-1 to finish): "))
    if score == -1:          
        break
    if score < 0 or score > 100:
        print("Invalid score. Enter a value between 0 and 100.")
        continue
    count += 1
    total += score
    if score > highest:
        highest = score
    if score < lowest:
        lowest = score

if count > 0:
    print("Students:", count)
    print("Average:", format(total / count, ".2f"))   
    print("Highest:", highest)
    print("Lowest:", lowest)
else:
    print("No scores were entered.")
````

### Question 17: Supermarket Inventory Restock Alert

**Goal:**  
Monitor product stock levels in a supermarket. The program keeps asking for the current stock of different items and alerts the manager when stock falls below a threshold.

**Requirements:**  
- Hard-code a minimum threshold of 10 units.  
- Use a `while True` loop.  
- In each iteration ask for: product name and current quantity.  
- If quantity < 10, print a restock alert for that product.  
- If the user enters `"done"` as the product name, exit the loop.  
- After exiting, print how many products triggered an alert.

**Concepts:** Infinite loop + sentinel, counters, conditional alerts.

**Sample Interaction:**
```
Product name (or 'done'): Notebooks
Quantity: 7
ALERT: Notebooks stock is low (7 units).
Product name (or 'done'): Pens
Quantity: 25
Product name (or 'done'): done
Total alerts issued: 1
```


````python
threshold = 10
alerts = 0

while True:
    product = input("Product name (or 'done'): ")
    if product == "done":
        break
    quantity = int(input("Quantity: "))
    if quantity < threshold:
        print("ALERT:", product, "stock is low (" + str(quantity) + " units).")
        alerts += 1

print("Total alerts issued:", alerts)
````

### Question 19: City Bus Ticket Booking System

**Goal:**  
Simulate a small city bus with a limited number of seats. Passengers can book seats until the bus is full or the booking window is closed.

**Requirements:**  
- The bus has 8 seats. Keep a counter of booked seats (start at 0).  
- Use a `while` loop that continues while seats are still available.  
- In each iteration:  
  - Ask the passenger for their name.  
  - Confirm the booking and increase the booked-seat counter.  
  - Show remaining seats.  
- After the loop (bus is full), print a message that the bus is now full and list how many passengers were boarded.  
- **Bonus challenge:** Allow an early exit if the user types `"close"` as the name; in that case print that booking is closed early.

**Concepts:** Counter-controlled loop, early exit with break, state tracking.

**Sample Interaction:**
```
Enter passenger name: Ali
Seat booked for Ali. Seats left: 7
Enter passenger name: Sara
Seat booked for Sara. Seats left: 6
...
(continues until 8 bookings or 'close')
```


````python
total_seats = 8
booked = 0
closed_early = False

while booked < total_seats:
    name = input("Enter passenger name: ")
    if name.lower() == "close":      
        closed_early = True
        break
    booked += 1
    print("Seat booked for", name + ".", "Seats left:", total_seats - booked)

if closed_early:
    print("Booking closed early.")
else:
    print("The bus is now full.")
print("Passengers boarded:", booked)
````
